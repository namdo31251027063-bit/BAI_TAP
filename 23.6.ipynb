```python
import osmnx as ox
import networkx as nx
import folium
ox.settings.log_console = True
ox.settings.use_cache = True
# Define a place that encompasses both Ben Thanh Market and Saigon Centre
place = "District 1, Ho Chi Minh City, Vietnam"
G_walk = ox.graph_from_place(place, network_type='walk')
# Get the coordinates of Ben Thanh Market and Saigon Centre using geocoding
origin_coords = ox.geocode("Ben Thanh Market, Ho Chi Minh City, Vietnam")
destination_coords = ox.geocode("Saigon Centre, Ho Chi Minh City, Vietnam")
# Get the nearest nodes in the downloaded graph to the geocoded coordinates
orig_node = ox.distance.nearest_nodes(G_walk, origin_coords[1], origin_coords[0])
dest_node = ox.distance.nearest_nodes(G_walk, destination_coords[1],destination_coords[0])
# Tính độ dài trung bình
lengths = [data["length"] for u, v, data in G_walk.edges(data=True)]
avg_length = sum(lengths) / len(lengths)
print(f"Số nút giao: {orig_node}")
print(f"Số đoạn đường: {dest_node}")
print(f"Độ dài trung bình: {avg_length:.2f} m")

# Calculate center_coords
center_coords = [(origin_coords[0] + destination_coords[0]) / 2, (origin_coords[1] + destination_coords[1]) / 2]

# Tạo bản đồ
m = folium.Map(location=center_coords, zoom_start=12)
# Get graph edges as a GeoDataFrame for plotting
edges = ox.graph_to_gdfs(G_walk, nodes=False)
# Vẽ đường
for _, row in edges.iterrows():
    if 'geometry' in row and row['geometry'] is not None:
        # Folium PolyLine expects coordinates in (latitude, longitude) format
        # Shapely LineString.coords gives (longitude, latitude)
        coords = [(lat, lon) for lon, lat in row['geometry'].coords]
        folium.PolyLine(coords, color='blue', weight=2, opacity=0.8).add_to(m)
m
```
