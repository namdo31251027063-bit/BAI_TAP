```python
import folium
from geopy.geocoders import Nominatim

# Khởi tạo
geolocator = Nominatim(user_agent="geoapi")

# Điểm trung tâm
center = geolocator.geocode("Ben Thanh Market, Ho Chi Minh City")
center_coords = (center.latitude, center.longitude)

# Tạo bản đồ
m = folium.Map(location=center_coords, zoom_start=12)

# Marker trung tâm
folium.Marker(center_coords,popup="Trung tâm: Bến Thành",icon=folium.Icon(color="red")).add_to(m)

# Vẽ các vùng phục vụ
# 3 km
folium.Circle(location=center_coords,radius=3000,color="green",fill=True,fill_opacity=0.2,popup="Vùng 3km").add_to(m)

# 5 km
folium.Circle(location=center_coords,radius=5000,color="orange",fill=True,fill_opacity=0.2,popup="Vùng 5km").add_to(m)

# 10 km
folium.Circle(location=center_coords,radius=10000,color="red",fill=True,fill_opacity=0.2,popup="Vùng 10km").add_to(m)

m
```
