```python
import folium
from geopy.geocoders import Nominatim
from geopy.distance import geodesic

# Khởi tạo
geolocator = Nominatim(user_agent="geoapi")

# Điểm trung tâm (chọn Bến Thành)
center = geolocator.geocode("Ben Thanh Market, Ho Chi Minh City")
center_coords = (center.latitude, center.longitude)

# Danh sách địa điểm
places = [
    "Đại học Kinh Tế Thành Phố Hồ Chí Minh cơ sở a, Ho Chi Minh City, Vietnam",
    "Đại học Kinh Tế Thành Phố Hồ Chí Minh cơ sở b, Ho Chi Minh City, Vietnam",
    "Đại học Kinh Tế Thành Phố Hồ Chí Minh cơ sở c, Ho Chi Minh City, Vietnam",
    "Đại học Kinh Tế Thành Phố Hồ Chí Minh cơ sở d, Ho Chi Minh City, Vietnam",
    "Phố đi bộ Nguyễn Huệ, Ho Chi Minh City, Vietnam",
    "Bách Khoa university, Ho Chi Minh City, Vietnam",
    "Nhà thờ Đức Bà, Ho Chi Minh",
    "Dinh Độc Lập, Ho Chi Minh",
    "Landmark 81, Ho Chi Minh"
]

# Tạo bản đồ
m = folium.Map(location=center_coords, zoom_start=13)

# Marker trung tâm
folium.Marker(center_coords,popup="Trung tâm: Bến Thành",icon=folium.Icon(color="red")).add_to(m)

# Duyệt từng địa điểm
for place in places:
    location = geolocator.geocode(place)

    if location:
        coords = (location.latitude, location.longitude)

        # Tính khoảng cách
        distance = geodesic(center_coords, coords).km

        # Marker
        folium.Marker(coords,popup=f"{place}\nKhoảng cách: {distance:.2f} km",icon=folium.Icon(color="blue")).add_to(m)

m
```
