```python
import folium
from geopy.geocoders import Nominatim
from geopy.distance import geodesic

# Khởi tạo
geolocator = Nominatim(user_agent="geoapi")

# Điểm trung tâm (chọn Bến Thành)
center = geolocator.geocode("Ben Thanh Market, Ho Chi Minh City")
center_coords = (center.latitude, center.longitude)

# Danh sách địa điểm
places = [
    "Đại học Kinh Tế Thành Phố Hồ Chí Minh cơ sở a, Ho Chi Minh City, Vietnam",
    "Đại học Kinh Tế Thành Phố Hồ Chí Minh cơ sở b, Ho Chi Minh City, Vietnam",
    "Đại học Kinh Tế Thành Phố Hồ Chí Minh cơ sở c, Ho Chi Minh City, Vietnam",
    "Đại học Kinh Tế Thành Phố Hồ Chí Minh cơ sở d, Ho Chi Minh City, Vietnam",
    "Phố đi bộ Nguyễn Huệ, Ho Chi Minh City, Vietnam",
    "Bách Khoa university, Ho Chi Minh City, Vietnam",
    "Nhà thờ Đức Bà, Ho Chi Minh",
    "Dinh Độc Lập, Ho Chi Minh",
    "Landmark 81, Ho Chi Minh"
]

# Tạo bản đồ
m = folium.Map(location=center_coords, zoom_start=13)

# Marker trung tâm
folium.Marker(center_coords,popup="Trung tâm: Bến Thành",icon=folium.Icon(color="red")).add_to(m)

# Duyệt từng địa điểm
for place in places:
    location = geolocator.geocode(place)

    if location:
        coords = (location.latitude, location.longitude)

        # Tính khoảng cách
        distance = geodesic(center_coords, coords).km

        # Marker
        folium.Marker(coords,popup=f"{place}\nKhoảng cách: {distance:.2f} km",icon=folium.Icon(color="blue")).add_to(m)

m
```

```python
import osmnx as ox
import networkx as nx
import folium
ox.settings.log_console = True
ox.settings.use_cache = True
# Define a place that encompasses both Ben Thanh Market and Saigon Centre
place = "District 1, Ho Chi Minh City, Vietnam"
G_walk = ox.graph_from_place(place, network_type='walk')
# Get the coordinates of Ben Thanh Market and Saigon Centre using geocoding
origin_coords = ox.geocode("Ben Thanh Market, Ho Chi Minh City, Vietnam")
destination_coords = ox.geocode("Saigon Centre, Ho Chi Minh City, Vietnam")
# Get the nearest nodes in the downloaded graph to the geocoded coordinates
orig_node = ox.distance.nearest_nodes(G_walk, origin_coords[1], origin_coords[0])
dest_node = ox.distance.nearest_nodes(G_walk, destination_coords[1],destination_coords[0])
# Tính độ dài trung bình
lengths = [data["length"] for u, v, data in G_walk.edges(data=True)]
avg_length = sum(lengths) / len(lengths)
print(f"Số nút giao: {orig_node}")
print(f"Số đoạn đường: {dest_node}")
print(f"Độ dài trung bình: {avg_length:.2f} m")

# Calculate center_coords
center_coords = [(origin_coords[0] + destination_coords[0]) / 2, (origin_coords[1] + destination_coords[1]) / 2]

# Tạo bản đồ
m = folium.Map(location=center_coords, zoom_start=12)
# Get graph edges as a GeoDataFrame for plotting
edges = ox.graph_to_gdfs(G_walk, nodes=False)
# Vẽ đường
for _, row in edges.iterrows():
    if 'geometry' in row and row['geometry'] is not None:
        # Folium PolyLine expects coordinates in (latitude, longitude) format
        # Shapely LineString.coords gives (longitude, latitude)
        coords = [(lat, lon) for lon, lat in row['geometry'].coords]
        folium.PolyLine(coords, color='blue', weight=2, opacity=0.8).add_to(m)
m
```

```python
import folium
from geopy.geocoders import Nominatim
import osmnx as ox
import networkx as nx
# Locations
address_A = "Đại học Kinh Tế Thành Phố Hồ Chí Minh cơ sở a, Ho Chi Minh City, Vietnam"
address_B = "Đại học Kinh Tế Thành Phố Hồ Chí Minh cơ sở b, Ho Chi Minh City, Vietnam"
# Get coordinates
geolocator = Nominatim(user_agent="routing-example", timeout=10) # Added timeout
location_A = geolocator.geocode(address_A)
location_B = geolocator.geocode(address_B)
# Check if locations were found before proceeding
latlng_A = (location_A.latitude, location_A.longitude)
latlng_B = (location_B.latitude, location_B.longitude)
# Download road network (graph) from OpenStreetMap
# Note: we take the road network in the area between A and B, radius about 3km
midpoint = ((latlng_A[0] + latlng_B[0]) / 2, (latlng_A[1] + latlng_B[1]) / 2)
G = ox.graph_from_point(midpoint, dist=3000, network_type='drive')
# Find the closest node to A and B
orig_node = ox.distance.nearest_nodes(G, latlng_A[1], latlng_A[0])
dest_node = ox.distance.nearest_nodes(G, latlng_B[1], latlng_B[0])
# Find the shortest path by length
route = nx.shortest_path(G, orig_node, dest_node, weight='length')
# Route coordinates
route_coords = [(G.nodes[n]['y'], G.nodes[n]['x']) for n in route]
# Create a map
m = folium.Map(location=midpoint, zoom_start=14)
# Add markers A and B
folium.Marker(latlng_A, popup="UEH cơ sở A", icon=folium.Icon(color='red')).add_to(m)
folium.Marker(latlng_B, popup="UEH cơ sở B", icon=folium.Icon(color='blue')).add_to(m)
# Draw the route according to reality
folium.PolyLine(locations=route_coords, color='green', weight=5, opacity=0.8).add_to(m)
m
```
